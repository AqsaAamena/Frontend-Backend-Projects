Q1. (Dataset Selection & Preprocessing)
Choose one classification dataset and one regression dataset (you can use
previous datasets or find new ones).
Load and clean both datasets.
Perform necessary preprocessing (encoding, scaling, train-test split).
Clearly identify Independent Features (X) and Dependent Feature (y)

```python
#Classification dataset

import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
clas_data= pd.read_csv('breast-cancer.csv')
# Drop ID column
clas_data= clas_data.drop("id", axis=1)
# Encode target
clas_data["diagnosis"] = clas_data["diagnosis"].map({"M":0, "B":1})
# Define features and target
X_clf= clas_data.drop("diagnosis", axis=1)   # Independent features
y = clas_data["diagnosis"]                # Dependent feature
# Scale features
clf_scaler = StandardScaler()
scaled_clas= clf_scaler.fit_transform(X_clf)
# Train-test split
X_train_clas, X_test_clas, y_train_clas, y_test_clas = train_test_split(
    scaled_clas, y, test_size=0.2, random_state=42
)
```

```python
#Regression Dataset
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
reg_data = pd.read_csv("medical-charges.csv")
# Encode categorical variables
reg_data = pd.get_dummies(reg_data, columns=["sex", "smoker", "region"], drop_first=True)
# Define features and target
X_reg = reg_data.drop("charges", axis=1)   # Independent features
y = reg_data["charges"]                # Dependent feature
# Scale features
reg_scaler = StandardScaler()
scaled_reg= reg_scaler.fit_transform(X_reg)
# Train-test split
X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(
    scaled_reg, y, test_size=0.2, random_state=42
)
```

Q2. (Classification Algorithms)
On the classification dataset, train and evaluate the following models:
Logistic Regression
Decision Tree Classifier
Support Vector Machine (SVM)
K-Nearest Neighbors (KNN)
Naive Bayes
For each model:
Train the model
Make predictions
Calculate Accuracy, Confusion Matrix, and Classification Report
Create a comparison table of all models’ performance.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
# Define models
classification_models = {
    "Logistic Regression": LogisticRegression(max_iter=1000),
    "Decision Tree": DecisionTreeClassifier(random_state=42),
    "SVM": SVC(kernel="linear", random_state=42),
    "KNN (k=5)": KNeighborsClassifier(n_neighbors=5),
    "Naive Bayes": GaussianNB()
}
results = []
# Train and evaluate each model
for name, model in classification_models.items():
    model.fit(X_train_clas, y_train_clas)
    y_pred = model.predict(X_test_clas)
    acc = accuracy_score(y_test_clas, y_pred)
    cm = confusion_matrix(y_test_clas, y_pred)
    report = classification_report(y_test_clas, y_pred, target_names=["Malignant","Benign"])
    print(f"\n{name}")
    print("Confusion Matrix:\n", cm)
    print("Classification Report:\n", report)
    results.append([name, acc])
# Comparison table
df_results = pd.DataFrame(results, columns=["Algorithm", "Accuracy"])
print("\nComparison Table:\n", df_results)
```

On the regression dataset, train and evaluate the following models:
Linear Regression
Decision Tree Regressor

Support Vector Regressor (SVR)
K-Nearest Neighbors Regressor
For each model:
Train the model
Make predictions
Calculate R2 Score
Create a comparison table showing which model performed best.

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.svm import SVR
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics import r2_score

# Define models
Regression_models = {
    "Linear Regression": LinearRegression(),
    "Decision Tree Regressor": DecisionTreeRegressor(random_state=42),
    "Support Vector Regressor": SVR(kernel="rbf"),
    "KNN Regressor (k=5)": KNeighborsRegressor(n_neighbors=5)
}

results = []

# Train and evaluate each model
for name, model in Regression_models.items():
    model.fit(X_train_reg, y_train_reg)
    y_pred = model.predict(X_test_reg)
    r2 = r2_score(y_test_reg, y_pred)
    results.append([name, r2])
    print(f"{name} R² Score: {r2:.4f}")

# Comparison table
df_results = pd.DataFrame(results, columns=["Algorithm", "R² Score"])
print("\nComparison Table:\n", df_results)
```

Q4. (Best Model Selection & Saving)
From Q2 and Q3:
Select the best classification model and best regression model based on
performance metrics.
Save both best models using joblib (also save scaler and columns if used).

```python
import joblib

# Save best classification model (Naive Bayes)
best_clf = GaussianNB()
best_clf.fit(X_train_clas, y_train_clas)
joblib.dump(clf_scaler, "clf_scaler.pkl")
joblib.dump(X_clf.columns.tolist(), "clf_columns.pkl")
joblib.dump(best_clf, "best_classification_model.pkl")

# Save best regression model (KNN Regressor)
best_reg = KNeighborsRegressor(n_neighbors=5)
best_reg.fit(X_train_reg, y_train_reg)  
joblib.dump(reg_scaler, "reg_scaler.pkl")
joblib.dump(X_reg.columns.tolist(), "reg_columns.pkl")
joblib.dump(best_reg, "best_regression_model.pkl")
```
