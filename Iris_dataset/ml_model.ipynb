Q1. (Manual Hyperparameter Tuning – KNN)
Load the Iris dataset.
Split into train and test sets (test_size=0.33, random_state=42).
Create a KNN model and manually try different values of n_neighbors (e.g., 3,
5, 7, 11, 13, 15).
For each value, train the model and print the test score.
Write which value gave the highest accuracy.

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
#Load dataset
iris = load_iris()
X, y = iris.data, iris.target
#Split into train and test sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.33, random_state=42 )
#Try different values of n_neighbors
neighbors_list = [3, 5, 7, 11, 13, 15]
scores = {}
for k in neighbors_list:
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train, y_train)
    score = knn.score(X_test, y_test)
    scores[k] = score
    print(f"n_neighbors={k}, Test Accuracy={score:.4f}")
#Find the best k
best_k = max(scores, key=scores.get)
print(f"\nBest accuracy achieved with n_neighbors={best_k}, Accuracy={scores[best_k]:.4f}")
```

Q2. (Manual Hyperparameter Tuning – SVM)
Using the same train-test split from Q1:
Create an SVM model and manually try different combinations of C (1, 10, 20)
and kernel (linear, rbf).
Train each combination and print the test accuracy.
Identify the best combination of hyperparameters.

```python
from sklearn.svm import SVC
# Hyperparameter combinations
C_values = [1, 10, 20]
kernels = ['linear', 'rbf']
results = {}
for C in C_values:
    for kernel in kernels:
        svm = SVC(C=C, kernel=kernel, random_state=42)
        svm.fit(X_train, y_train)
        score = svm.score(X_test, y_test)
        results[(C, kernel)] = score
        print(f"C={C}, kernel={kernel}, Test Accuracy={score:.4f}")
# Find best combination
best_params = max(results, key=results.get)
print(f"\nBest combination: C={best_params[0]}, kernel={best_params[1]}, "
      f"Accuracy={results[best_params]:.4f}")
```

Q3. (Grid Search CV)
Apply GridSearchCV on an SVM model:
Create a DataFrame of cv_results_ and display param_C, param_kernel, and
mean_test_score.
Print the best parameters and best score.

```python
import pandas as pd
from sklearn.model_selection import GridSearchCV
from sklearn.svm import SVC
# Define parameter grid
param_grid = {
    'C': [1, 10, 20],
    'kernel': ['linear', 'rbf']
}
# Create GridSearchCV
grid = GridSearchCV(SVC(random_state=42), param_grid, cv=5, return_train_score=True)
grid.fit(X_train, y_train)
# Convert results to DataFrame
results_df = pd.DataFrame(grid.cv_results_)
# Display selected columns
print(results_df[['param_C', 'param_kernel', 'mean_test_score']])
# Print best parameters and best score
print("\nBest Parameters:", grid.best_params_)
print("Best Score:", grid.best_score_)
```

Q4. (Randomized Search CV)
Apply RandomizedSearchCV on the same SVM model:
Use the same parameter grid as Q3.
Set n_iter=5 and cv=5.
Fit the model.
Display the results DataFrame with important columns.
Compare the best parameters and score with Grid Search results from Q3.

```python
from sklearn.model_selection import RandomizedSearchCV
from sklearn.svm import SVC
param_grid = {
    'C': [1, 10, 20],
    'kernel': ['linear', 'rbf']}
# Create RandomizedSearchCV
random_search = RandomizedSearchCV(
    SVC(random_state=42),
    param_distributions=param_grid,
    n_iter=5,
    cv=5,   
    random_state=42,)
random_search.fit(X_train, y_train)
# Convert results to DataFrame
rand_results_df = pd.DataFrame(random_search.cv_results_)
# Display important columns
print(rand_results_df[['param_C', 'param_kernel', 'mean_test_score']])
# Print best parameters and best score
print("\nBest Parameters:", random_search.best_params_)
print("Best Score:", random_search.best_score_)
```

Q5. (Bagging – Random Forest)
Train a Random Forest Classifier on the Iris dataset:
Use n_estimators=100 and random_state=42.
Perform train-test split (test_size=0.2, stratify=y).
Train the model and calculate accuracy on the test set.
Print the accuracy score.

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
#Train-test split (stratified)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)
#Train Random Forest
rf = RandomForestClassifier(n_estimators=100, random_state=42)
rf.fit(X_train, y_train)
#Predict and calculate accuracy
y_pred = rf.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print(f"Test Accuracy: {accuracy:.4f}")
```

Q6. (Boosting – AdaBoost & Gradient Boosting)
Train the following models:
AdaBoostClassifier (n_estimators=100, random_state=42)
GradientBoostingClassifier (n_estimators=100, learning_rate=0.1,
random_state=42)
Calculate and print the accuracy of both models on the test set.

```python
from sklearn.ensemble import AdaBoostClassifier, GradientBoostingClassifier
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
# Train-test split (stratified)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)
# AdaBoost Classifier
ada = AdaBoostClassifier(n_estimators=100, random_state=42)
ada.fit(X_train, y_train)
y_pred_ada = ada.predict(X_test)
ada_acc = accuracy_score(y_test, y_pred_ada)
#Gradient Boosting Classifier
gb = GradientBoostingClassifier(n_estimators=100, learning_rate=0.1, random_state=42)
gb.fit(X_train, y_train)
y_pred_gb = gb.predict(X_test)
gb_acc = accuracy_score(y_test, y_pred_gb)
print(f"AdaBoost Test Accuracy: {ada_acc:.4f}")
print(f"GradientBoosting Test Accuracy: {gb_acc:.4f}")
```

Q7. (Boosting – XGBoost)
Train an XGBClassifier with:
n_estimators=100
learning_rate=0.1
max_depth=3
random_state=42
Calculate and print the test accuracy.

```python
from xgboost import XGBClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

# Train-test split (stratified)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)
# Train XGBClassifier
xgb = XGBClassifier(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=3,
    random_state=42,
    use_label_encoder=False,
    eval_metric='mlogloss'   # avoids warning
)
xgb.fit(X_train, y_train)
#Predict and calculate accuracy
y_pred = xgb.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print(f"XGBoost Test Accuracy: {accuracy:.4f}")
```

Q8. (Hyperparameter Tuning on Random Forest)
Apply GridSearchCV on Random Forest Classifier:
Parameter grid:
n_estimators: [50, 100, 150]
max_depth: [3, 5, 7]
Use cv=5.
Find and print the best parameters and best score.

```python
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.ensemble import RandomForestClassifier
#Train-test split (stratified)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)
#Define parameter grid
param_grid = {
    'n_estimators': [50, 100, 150],
    'max_depth': [3, 5, 7]
}
#GridSearchCV
grid = GridSearchCV(
    RandomForestClassifier(random_state=42),
    param_grid,
    cv=5,
    return_train_score=True
)
grid.fit(X_train, y_train)
#Results DataFrame
results_df = pd.DataFrame(grid.cv_results_)
print(results_df[['param_n_estimators', 'param_max_depth', 'mean_test_score']])
#Best parameters and score
print("\nBest Parameters:", grid.best_params_)
print("Best Score:", grid.best_score_)
```

Q9. (Complete Model Comparison)
On the Iris dataset (or any other classification dataset):
Train and evaluate the following models:
SVM (with best parameters from GridSearch/RandomSearch)
Random Forest
AdaBoost
Gradient Boosting
XGBoost
Create a comparison table of all models with their

```python
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier, AdaBoostClassifier, GradientBoostingClassifier
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score
#Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
    )
#Define models
models = {
    "SVM (linear kernel)": SVC(C=1, kernel='linear', random_state=42),
    "Random Forest": RandomForestClassifier(n_estimators=100, random_state=42),
    "AdaBoost": AdaBoostClassifier(n_estimators=100, random_state=42),
    "Gradient Boosting": GradientBoostingClassifier(n_estimators=100, learning_rate=0.1, random_state=42),
    "XGBoost": XGBClassifier(n_estimators=100, learning_rate=0.1, max_depth=3,
                             random_state=42, use_label_encoder=False, eval_metric='mlogloss')
}
results = {}
for name, model in models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    acc = accuracy_score(y_test, y_pred)
    results[name] = acc
#Create comparison table
comparison_df = pd.DataFrame(list(results.items()), columns=["Model", "Test Accuracy"])
print(comparison_df)
#Best model
best_model = max(results, key=results.get)
print(f"\nBest Model: {best_model}, Accuracy={results[best_model]:.4f}")
```

Q10. (Mini Project – Full Pipeline)
Choose any classification dataset.
1. Perform train-test split.
2. Apply Manual Search, GridSearchCV, and RandomizedSearchCV on SVM or
KNN.
3. Train Random Forest and at least two Boosting models (AdaBoost, Gradient
Boosting, XGBoost).
4. Compare all results in a table.
5. Save the best model using joblib.
6. Create a Frontend for the model.

```python
from sklearn.model_selection import train_test_split, GridSearchCV, RandomizedSearchCV
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier, AdaBoostClassifier, GradientBoostingClassifier
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score
import joblib
# 1.train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)
# 3. Manual Search (KNN example)
manual_scores = {}
for k in [3, 5, 7, 11]:
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train, y_train)
    manual_scores[k] = accuracy_score(y_test, knn.predict(X_test))
print("Manual KNN Scores:", manual_scores)
# GridSearchCV (SVM example)
param_grid = {'C': [1, 10, 20], 'kernel': ['linear', 'rbf']}
grid = GridSearchCV(SVC(random_state=42), param_grid, cv=5)
grid.fit(X_train, y_train)
print("GridSearch Best Params:", grid.best_params_, "Best Score:", grid.best_score_)
# RandomizedSearchCV (SVM example)
rand = RandomizedSearchCV(SVC(random_state=42), param_grid, n_iter=5, cv=5, random_state=42)
rand.fit(X_train, y_train)
print("RandomSearch Best Params:", rand.best_params_, "Best Score:", rand.best_score_)
# 4. Train ensemble models
models = {
    "SVM (best)": SVC(C=grid.best_params_['C'], kernel=grid.best_params_['kernel'], random_state=42),
    "Random Forest": RandomForestClassifier(n_estimators=100, random_state=42),
    "AdaBoost": AdaBoostClassifier(n_estimators=100, random_state=42),
    "Gradient Boosting": GradientBoostingClassifier(n_estimators=100, learning_rate=0.1, random_state=42),
    "XGBoost": XGBClassifier(n_estimators=100, learning_rate=0.1, max_depth=3,
                             random_state=42,)
}
results = {}
for name, model in models.items():
    model.fit(X_train, y_train)
    acc = accuracy_score(y_test, model.predict(X_test))
    results[name] = acc
# 5. Comparison table
comparison_df = pd.DataFrame(list(results.items()), columns=["Model", "Test Accuracy"])
print("\nModel Comparison:\n", comparison_df)
#Save best model
best_model_name = max(results, key=results.get)
best_model = models[best_model_name]
joblib.dump(best_model, "best_model.pkl")
print(f"\nBest model saved: {best_model_name}")
```
